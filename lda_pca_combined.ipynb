```python
import shutil
from pathlib import Path
from tqdm import tqdm

# ---------------- CONFIG ----------------
PCA_ROOT = Path("dataset_pca3")
LDA_ROOT = Path("dataset_lda_pca")
OUT_ROOT = Path("dataset_combined")
SPLITS   = ["train", "valid", "test"]
# ----------------------------------------

for split in SPLITS:
    out_img = OUT_ROOT / "images" / split
    out_lbl = OUT_ROOT / "labels" / split
    out_img.mkdir(parents=True, exist_ok=True)
    out_lbl.mkdir(parents=True, exist_ok=True)

    pca_images = list((PCA_ROOT / "images" / split).glob("*.png"))

    for img_path in tqdm(pca_images, desc=split):
        stem = img_path.stem
        lbl_name = stem + ".txt"

        # --- PCA-3 image + label ---
        shutil.copy(img_path, out_img / f"pca_{img_path.name}")

        pca_lbl = PCA_ROOT / "labels" / split / lbl_name
        if pca_lbl.exists():
            shutil.copy(pca_lbl, out_lbl / f"pca_{lbl_name}")

        # --- LDA+PCA image + label ---
        lda_img = LDA_ROOT / "images" / split / img_path.name
        if lda_img.exists():
            shutil.copy(lda_img, out_img / f"lda_{img_path.name}")

        lda_lbl = LDA_ROOT / "labels" / split / lbl_name
        if lda_lbl.exists():
            shutil.copy(lda_lbl, out_lbl / f"lda_{lbl_name}")

    print(f"  {split}: {len(list(out_img.glob('*.png')))} images | {len(list(out_lbl.glob('*.txt')))} labels")

print(f"\n✅ Done. Combined dataset saved to: {OUT_ROOT}")
```

```python
from ultralytics import YOLO

print("Starting YOLO training with 4-channel input...")

model = YOLO("yolo11n.yaml")

model.train(
    data="weighted_fusion.yaml",
    imgsz=1024,
    epochs=500,
    patience=80,        # early stopping
    batch=8,
    device=0,
    project="weighted_3pca",
    name="weighted_yolo11n_pca3",
    amp=False,
    augment=False,
    workers=0,
)
```
